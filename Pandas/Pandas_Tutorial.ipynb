```python
#feature engineering 
df2=df1.copy()
df2["tax"]=df2["income"].apply(lambda x: "20%" if x >=500 else "10%")
df2
df2_dic={"Male" : "m" , "Female": "f", "NaN":"O"}
df2["gender"]=df2["gender"].map(df2_dic)
df2

df3=df2.assign(new_income =df2["income"] * 1.1)
df3

df2["country"]=df2["country"].replace("USA" ,"INDIA")
df2
```

```python
clean_data=df1.copy()
df1_mean=clean_data["age"].mean()
clean_data["age"]=clean_data["age"].fillna(df1_mean)
clean_data['age'] = clean_data['age'].astype(int)

clean_data
```

```python
#melt & Pivot

df=pd.DataFrame({
    "country" : ["usa","usa","india","india"],
    "year" : [2020,2021,2020,2021],
    "sales": [100,120,90,110],
    "profit":[20,25,18,22],
    "tax":[20,20,10,20]
})
melted_df= df.melt(
    id_vars=["country","year"],
    value_vars=["sales","profit","tax"],
    var_name="metrics",
    value_name="value"
)
original=melted_df.pivot(
    index=["country","year"],
    columns="matrics",
    values="values"
)
original
```
